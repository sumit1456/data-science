```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

# Column names for Iris dataset
columns = [
    'SepalLengthCm',
    'SepalWidthCm',
    'PetalLengthCm',
    'PetalWidthCm',
    'Species'
]

# Read iris.data
df = pd.read_csv('iris/iris.data', names=columns)

# Remove empty rows, if any
df.dropna(inplace=True)

# Display dataset
print("Dataset:")
print(df.head())

# Convert numerical values into categorical ranges
df['SepalLength'] = pd.cut(
    df['SepalLengthCm'],
    bins=3,
    labels=['Short', 'Medium', 'Long']
)

df['SepalWidth'] = pd.cut(
    df['SepalWidthCm'],
    bins=3,
    labels=['Narrow', 'Medium', 'Wide']
)

df['PetalLength'] = pd.cut(
    df['PetalLengthCm'],
    bins=3,
    labels=['Short', 'Medium', 'Long']
)

df['PetalWidth'] = pd.cut(
    df['PetalWidthCm'],
    bins=3,
    labels=['Narrow', 'Medium', 'Wide']
)

# Select categorical columns
data = df[
    ['SepalLength', 'SepalWidth',
     'PetalLength', 'PetalWidth', 'Species']
]

# Convert categorical data into one-hot encoded format
basket = pd.get_dummies(data)

# Apply Apriori algorithm
frequent_itemsets = apriori(
    basket,
    min_support=0.1,
    use_colnames=True
)

print("\nFrequent Itemsets:")
print(frequent_itemsets)

# Generate association rules
rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.6
)

print("\nAssociation Rules:")
print(rules[
    ['antecedents', 'consequents',
     'support', 'confidence', 'lift']
])
```
