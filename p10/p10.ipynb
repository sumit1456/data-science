```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage

# 1. Read the dataset
df = pd.read_csv("Wholesale customers data.csv")

# 2. Display the dataset
print("Dataset:")
print(df.head())

# 3. Select features for clustering
X = df[['Fresh', 'Milk', 'Grocery',
        'Frozen', 'Detergents_Paper', 'Delicassen']]

# 4. Standardize the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 5. Create linkage matrix
linked = linkage(X_scaled, method='ward')

# 6. Display dendrogram
plt.figure(figsize=(12, 6))

dendrogram(
    linked,
    truncate_mode='lastp',
    p=20
)

plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Customers")
plt.ylabel("Euclidean Distance")
plt.show()

# 7. Apply Hierarchical Clustering
model = AgglomerativeClustering(
    n_clusters=3,
    linkage='ward'
)

# 8. Predict cluster for each customer
clusters = model.fit_predict(X_scaled)

# 9. Add cluster information to dataset
df['Cluster'] = clusters

# 10. Display clustered data
print("\nClustered Dataset:")
print(df.head(10))

# 11. Display number of customers in each cluster
print("\nNumber of customers in each cluster:")
print(df['Cluster'].value_counts())
```
